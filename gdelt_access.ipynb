```python
with open('./Test CSV/header_2013-04-01.txt', 'r') as f:
    headers = f.read().strip().split('\t')

df = pd.read_csv('./Test CSV/20150218230000.tsv', delimiter='\t', header=None)
df.dataframeName = 'GDELT Test'
```

The data contains 61 columns, but the header text provided only contains 58 elements. While examining the data, it seems like columns at index 37 and 38, 45 and 46, 53 and 54 could have belong to the same column, but got split into two. The value of the earlier column was just the first two letters of the later column. We can just keep the later one and delete the earlier one to match the headers to the data.

```python
df = df.drop(df.columns[[37, 45, 53]], axis=1)
df.columns = headers
```
